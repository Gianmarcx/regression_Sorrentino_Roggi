```python
from sklearn.model_selection import train_test_split

from sklearn.linear_model import LinearRegression, Ridge, Lasso

from sklearn.preprocessing import StandardScaler

from sklearn.metrics import mean_squared_error as MSE

from sklearn.preprocessing import MinMaxScaler

from sklearn.metrics import r2_score

from sklearn.metrics import root_mean_squared_error as RMSE
```

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('california_housing_data.csv')
```

```python
pos_col = df.columns.tolist()
pos_col.remove("Longitude")
for col in pos_col:
    df = df[df[col] > 0 ]
```

```python
df["Rooms_per_person"] = df["AveRooms"] / df["AveOccup"]
df["Bedrooms_per_person"] = df["AveBedrms"] / df["AveOccup"]
df["BedRooms_per_room"] = df["AveBedrms"] / df["AveRooms"]
```

```python
features = df.columns.to_list()
features.remove("MedHouseVal")
X = df[features]
y = df["MedHouseVal"]
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=49)
```

```python
"""# inizializzo un mask di False per tutte le righe
mask = np.zeros(X_train.shape[0], dtype=bool)

for col in X_train.columns:
    Q1 = X_train[col].quantile(0.25)
    Q3 = X_train[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    # aggiorno il mask: True se la riga è un outlier in QUALSIASI colonna
    mask |= (X_train[col] < lower_bound) | (X_train[col] > upper_bound)

# applico il filtro una volta sola
X_train = X_train[~mask]
y_train = y_train[~mask]"""
```

```python
scaler = StandardScaler()
X_train_s = scaler.fit_transform(X_train)
X_test_s = scaler.transform(X_test)
```

```python
regression = LinearRegression()
regression.fit(X_train_s, y_train)
pred = regression.predict(X_test_s)

errore = MSE(y_test, pred)
r2 = r2_score(y_test, pred)

print("R2 Score:", r2)
print("Mean Squared Error:", errore)
```

```python
ridge_regression = Ridge(alpha=5.0)
ridge_regression.fit(X_train_s, y_train)
pred = ridge_regression.predict(X_test_s)

errore = MSE(y_test, pred)
r2 = r2_score(y_test, pred)

print("R2 Score:", r2)
print("Mean Squared Error:", errore)
print("Root Mean Squared Error:", RMSE(y_test, pred))
```

```python
lasso_regression = Lasso(alpha=0.1)
lasso_regression.fit(X_train_s, y_train)
pred = lasso_regression.predict(X_test_s)

errore = MSE(y_test, pred)
r2 = r2_score(y_test, pred)

print("R2 Score:", r2)
print("Mean Squared Error:", errore)
print("Root Mean Squared Error:", RMSE(y_test, pred))
```
